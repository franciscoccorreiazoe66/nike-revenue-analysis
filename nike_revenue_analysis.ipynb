```python
company= "Nike"
years= [2022,2023,2024,2025,2026]
print(company)
print(years)
revenues=[46710, 51217, 51362, 46309, 46398]
print(revenues)
print(len(revenues))
print(len(years))
print(f"{company} FY{years[0]} revenue: ${revenues[0]:,}M")
print(f"{company} FY{years[-1]} revenue: ${revenues[-1]:,}M")
total_revenue=sum(revenues)
print(f"{company} Total Revenue FY{years[0]}-{years[-1]}: ${total_revenue:,}M")
average_revenue= total_revenue/len(revenues)
print(f"{company} Average Revenue, FY{years[0]}-{years[-1]}: ${average_revenue:,.1f}M")
total_growth=revenues[-1]/revenues[0] -1
print(f"{company} Revenue Growth, FY{years[0]}-{years[-1]}: {total_growth:.1%}")
for i in range(len(years)):
    print(f"{company} FY{years[i]}: ${revenues[i]:,}M")
for i in range(1, len(years)):
    growth=revenues[i]/revenues[i-1]-1
    print(f"{company} FY{years[i]} Revenue Growth: {growth:.1%}")
periods=len(revenues)-1
cagr=(revenues[-1]/revenues[0]) ** (1/periods)-1
print(f"{company} CAGR, FY{years[0]}-{years[-1]}: {cagr:.1%}")
check=(1+cagr)**periods
print(check)
print(revenues[-1]/revenues[0])
def calculate_cagr(values):
    periods=len(values)-1
    cagr=(values[-1]/values[0])**(1/periods)-1
    return cagr
print(f"{company} CAGR: {calculate_cagr(revenues):.1%}")
print(f"Test List CAGR: {calculate_cagr([50,55,61,70]):.1%}")
import matplotlib.pyplot as plt
plt.bar(years,revenues)
plt.title(f"{company} Revenue")
plt.xlabel("Fiscal Year")
plt.ylabel("Revenue ($M)")
plt.xticks(years)
plt.show()











```
